```python
import pandas as pd
import matplotlib.pyplot as plt

def plot_training_results(results_path):
    """
    Plots the training loss, validation loss, accuracy, recall, and F1-score over epochs.

    Args:
        results_path (str): Path to the CSV file containing training logs.
    """

    # Load training logs
    df = pd.read_csv(results_path)

    # Create a figure with subplots
    fig, axes = plt.subplots(2, 2, figsize=(12, 10))

    # Plot Training and Validation Loss
    axes[0, 0].plot(df["Epoch"], df["Train Loss"], label="Train Loss", marker='o')
    axes[0, 0].plot(df["Epoch"], df["Val Loss"], label="Validation Loss", marker='s')
    axes[0, 0].set_title("Loss Over Epochs")
    axes[0, 0].set_xlabel("Epochs")
    axes[0, 0].set_ylabel("Loss")
    axes[0, 0].legend()
    axes[0, 0].grid()

    # Plot Accuracy
    axes[0, 1].plot(df["Epoch"], df["Train Accuracy"], label="Train Accuracy", marker='o')
    axes[0, 1].plot(df["Epoch"], df["Val Accuracy"], label="Validation Accuracy", marker='s')
    axes[0, 1].set_title("Accuracy Over Epochs")
    axes[0, 1].set_xlabel("Epochs")
    axes[0, 1].set_ylabel("Accuracy (%)")
    axes[0, 1].legend()
    axes[0, 1].grid()

    # Plot Recall
    axes[1, 0].plot(df["Epoch"], df["Train Recall"], label="Train Recall", marker='o')
    axes[1, 0].plot(df["Epoch"], df["Val Recall"], label="Validation Recall", marker='s')
    axes[1, 0].set_title("Recall Over Epochs")
    axes[1, 0].set_xlabel("Epochs")
    axes[1, 0].set_ylabel("Recall")
    axes[1, 0].legend()
    axes[1, 0].grid()

    # Plot F1-score
    axes[1, 1].plot(df["Epoch"], df["Train F1"], label="Train F1-score", marker='o')
    axes[1, 1].plot(df["Epoch"], df["Val F1"], label="Validation F1-score", marker='s')
    axes[1, 1].set_title("F1-score Over Epochs")
    axes[1, 1].set_xlabel("Epochs")
    axes[1, 1].set_ylabel("F1-score")
    axes[1, 1].legend()
    axes[1, 1].grid()

    # Adjust layout and show plot
    plt.tight_layout()
    plt.show()

plot_training_results("../src/results/single_model_training.csv")
```
